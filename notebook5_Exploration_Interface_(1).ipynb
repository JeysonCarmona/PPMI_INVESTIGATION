### Boolean Column Conversion

## 2. Flexible column detection

The notebook automatically searches for the real name of each expected column (in case of differences in capitalization, spaces, or aliases). If a column is not found, the interface simply omits that field/filter instead of failing.

```python
import unicodedata

def normalizar_nombre(s):
    """Normalizes a column name for comparison: no accents, lowercase,
    spaces/parentheses/hyphens converted to '_', no repeated '_'.
    """
    s = str(s).strip().lower()
    s = "".join(c for c in unicodedata.normalize("NFKD", s) if not unicodedata.combining(c))
    for ch in [" ", "(", ")", "-"]:
        s = s.replace(ch, "_")
    while "__" in s:
        s = s.replace("__", "_")
    return s.strip("_")


def detectar_columna(candidatos, columnas_df):
    """Searches for the actual column corresponding to a list of candidate names.
    1) Exact normalized match (ideal).
    2) If no exact match, coincidence by containment (e.g., 'n_estudios' fits
       within 'n_estudios_total'), to tolerate different suffixes/prefixes.
    """
    columnas_norm = {normalizar_nombre(c): c for c in columnas_df}
    candidatos_norm = [normalizar_nombre(c) for c in candidatos]

    for cand in candidatos_norm:
        if cand in columnas_norm:
            return columnas_norm[cand]

    for cand in candidatos_norm:
        for col_norm, col_real in columnas_norm.items():
            if cand in col_norm or col_norm in cand:
                return col_real

    return None


CANDIDATOS = {
    "patno":                ["PATNO"],
    "alias":                ["Alias (SI)", "Alias_SI", "Alias"],
    "sexo":                 ["Sexo"],
    "edad":                 ["ENROLL_AGE", "Edad"],
    "diagnostico":          ["COHORT_DEFINITION", "Diagnostico"],
    "estado":               ["ENROLL_STATUS", "Estado_clinico", "Estado (Enrolled)", "Estado"],
    "tiene_imagenes":       ["Tiene_imagenes"],
    "n_estudios":           ["N_estudios_total", "n_estudios"],
    "n_series":             ["N_series_total", "n_series"],
    "n_dicom":              ["N_archivos_dicom_total", "n_archivos_dicom"],
    "modalidades":          ["Modalidades_encontradas", "modalidades"],
    "orientaciones":        ["Orientaciones_encontradas", "orientaciones"],
    "secuencias":           ["Secuencias_encontradas", "secuencias"],
    "tiene_exoma":          ["Tiene_exoma"],
    "archivo_tar":          ["Archivo_TAR"],
    "archivo_vcf":          ["Ruta_interna_VCF", "archivo_vcf"],
    "ruta_primer_estudio":  ["Ruta_primer_estudio"],
    "ruta_primer_dicom":    ["Ruta_primer_dicom"],
    "ruta_interna_rar":     ["Ruta_interna_rar"],
    "paciente_completo":    ["paciente_completo"],
    "estado_investigacion": ["estado_investigacion"],
    "grupo_diagnostico_imagenes": ["Grupo_diagnostico_imagenes"],
    "cohort":               ["COHORT"],
    "enroll_date":          ["ENROLL_DATE"],
}

COL = {clave: detectar_columna(cands, df_master.columns) for clave, cands in CANDIDATOS.items()}

print("Detected column mapping:\n")
for clave, real in COL.items():
    estado = real if real else "NOT FOUND (this field/filter will be omitted)"
    print(f"  {clave:18s} -> {estado}")

assert COL["patno"] is not None, "Could not detect PATNO column. Check the master index."
```

### Patient Summary Card Generation
- `valor_o_na`: Helper function to get the value of a column for a given row, handling null values and numerical formats.
- `generar_ficha_html`: Generates a detailed HTML representation of a patient's information, formatted with styles and status badges, useful for interactive visualization.

```python
def col_bool(clave):
    """Converts a column that may come as text ('True'/'Sí') or bool to a real boolean."""
    real = COL.get(clave)
    if real is None or real not in df_master.columns:
        return pd.Series([False] * len(df_master), index=df_master.index)
    serie = df_master[real]
    if serie.dtype == bool:
        return serie
    return serie.astype(str).str.strip().str.lower().isin(["true", "sí", "si", "1", "yes"])

df_master["_tiene_imagenes_bool"] = col_bool("tiene_imagenes")
df_master["_tiene_exoma_bool"] = col_bool("tiene_exoma")

if COL.get("paciente_completo") is not None:
    df_master["_paciente_completo_bool"] = col_bool("paciente_completo")
else:
    # If the master index does not have the calculated column, it is derived here.
    df_master["_paciente_completo_bool"] = df_master["_tiene_imagenes_bool"] & df_master["_tiene_exoma_bool"]

print("Patients with images:", df_master["_tiene_imagenes_bool"].sum())
print("Patients with exome:", df_master["_tiene_exoma_bool"].sum())
print("Complete patients (images + exome):", df_master["_paciente_completo_bool"].sum())
```

## 3. Helper functions for search and summary card

```python
def opciones_filtro(clave, incluir_todos=True):
    """Returns the list of unique values available for a filter dropdown."""
    real = COL.get(clave)
    opciones = ["Todos"] if incluir_todos else []
    if real is None or real not in df_master.columns:
        return opciones
    valores = sorted(df_master[real].dropna().astype(str).unique().tolist())
    return opciones + valores


def buscar_pacientes(df, patno_texto="", alias_texto="", estado="Todos",
                      diagnostico="Todos", sexo="Todos", estado_investigacion="Todos",
                      solo_con_imagenes=False, solo_con_exoma=False):
    resultado = df.copy()

    if patno_texto.strip():
        resultado = resultado[resultado[COL["patno"]].astype(str).str.contains(patno_texto.strip(), na=False)]

    if alias_texto.strip() and COL.get("alias"):
        resultado = resultado[resultado[COL["alias"]].astype(str).str.contains(alias_texto.strip(), case=False, na=False)]

    if estado != "Todos" and COL.get("estado"):
        resultado = resultado[resultado[COL["estado"]].astype(str) == estado]

    if diagnostico != "Todos" and COL.get("diagnostico"):
        resultado = resultado[resultado[COL["diagnostico"]].astype(str) == diagnostico]

    if sexo != "Todos" and COL.get("sexo"):
        resultado = resultado[resultado[COL["sexo"]].astype(str) == sexo]

    if estado_investigacion != "Todos" and COL.get("estado_investigacion"):
        resultado = resultado[resultado[COL["estado_investigacion"]].astype(str) == estado_investigacion]

    if solo_con_imagenes:
        resultado = resultado[resultado["_tiene_imagenes_bool"]]

    if solo_con_exoma:
        resultado = resultado[resultado["_tiene_exoma_bool"]]

    return resultado


def columnas_tabla_resumen():
    """Columns to show in the results table (only those that exist)."""
    claves = ["patno", "alias", "edad", "sexo", "diagnostico", "estado",
              "estado_investigacion", "tiene_imagenes", "tiene_exoma"]
    return [COL[c] for c in claves if COL.get(c) is not None]
```

```python
def valor_o_na(fila, clave):
    real = COL.get(clave)
    if real is None or real not in fila.index or pd.isna(fila[real]):
        return "Not available"
    valor = fila[real]
    # Numbers like 170.0 are displayed as 170 (NaNs from a merge leave float even if they are counts)
    if isinstance(valor, float) and valor.is_integer():
        return int(valor)
    return valor


def generar_ficha_html(fila, modo_completo=False):
    """Generates the HTML for a patient's summary card."""
    patno = valor_o_na(fila, "patno")
    alias = valor_o_na(fila, "alias")
    estado_inv = valor_o_na(fila, "estado_investigacion")

    colores_estado = {
        "Imágenes + Exoma": "#2f855a",
        "Solo imágenes": "#2c5282",
        "Solo exoma": "#975a16",
        "Sin datos": "#718096",
    }
    color_badge = colores_estado.get(estado_inv, "#718096")

    html = f"""
    <div style="border:1px solid #ccc; border-radius:8px; padding:16px; font-family:sans-serif; max-width:650px;">
      <h3 style="margin-top:0;">Patient {patno} {'— ' + str(alias) if alias != 'Not available' else ''}
        <span style="font-size:0.55em; background:{color_badge}; color:white; padding:3px 10px; border-radius:12px; vertical-align:middle;">{estado_inv}</span>
      </h3>

      <h4 style="color:#2c5282;">Clinical Information</h4>
      <ul>
        <li><b>PATNO:</b> {patno}</li>
        <li><b>Alias (SI):</b> {alias}</li>
        <li><b>Age:</b> {valor_o_na(fila, 'edad')}</li>
        <li><b>Gender:</b> {valor_o_na(fila, 'sexo')}</li>
        <li><b>Diagnosis:</b> {valor_o_na(fila, 'diagnostico')}</li>
        <li><b>Status:</b> {valor_o_na(fila, 'estado')}</li>
      </ul>

      <h4 style="color:#2f855a;">Image Information</h4>
      <ul>
        <li><b>Has images:</b> {"Yes" if fila["_tiene_imagenes_bool"] else "No"}</li>
        <li><b>Number of studies:</b> {valor_o_na(fila, 'n_estudios')}</li>
        <li><b>Number of series:</b> {valor_o_na(fila, 'n_series')}</li>
        <li><b>Number of DICOM files:</b> {valor_o_na(fila, 'n_dicom')}</li>
        <li><b>Modalities:</b> {valor_o_na(fila, 'modalidades')}</li>
        <li><b>Orientations:</b> {valor_o_na(fila, 'orientaciones')}</li>
        <li><b>Sequences:</b> {valor_o_na(fila, 'secuencias')}</li>
        <li><b>Diagnostic group (by images):</b> {valor_o_na(fila, 'grupo_diagnostico_imagenes')}</li>
        <li><b>First study path:</b> {valor_o_na(fila, 'ruta_primer_estudio')}</li>
        <li><b>First DICOM path:</b> {valor_o_na(fila, 'ruta_primer_dicom')}</li>
        <li><b>Internal RAR path:</b> {valor_o_na(fila, 'ruta_interna_rar')}</li>
      </ul>

      <h4 style="color:#975a16;">Genetic Information</h4>
      <ul>
        <li><b>Has exome:</b> {"Yes" if fila["_tiene_exoma_bool"] else "No"}</li>
        <li><b>TAR file:</b> {valor_o_na(fila, 'archivo_tar')}</li>
        <li><b>Internal VCF path:</b> {valor_o_na(fila, 'archivo_vcf')}</li>
      </ul>
    """

    if modo_completo:
        html += """
      <div style="margin-top:12px; padding:10px; background:#ebf8ff; border-radius:6px; color:#2c5282;">
        This patient has clinical information, MRI studies, and exome data.
        Ready to be viewed in Notebook 6.
      </div>
        """

    html += "</div>"
    return html
```

## 4. Mode 1 — All Patients

```python
# --- Search and filter controls ---
m1_patno = widgets.Text(placeholder="Search by PATNO", description="PATNO:", layout=widgets.Layout(width="250px"))
m1_alias = widgets.Text(placeholder="Search by Alias (SI)", description="Alias:", layout=widgets.Layout(width="250px"))
m1_estado = widgets.Dropdown(options=opciones_filtro("estado"), description="Status:", layout=widgets.Layout(width="250px"))
m1_diagnostico = widgets.Dropdown(options=opciones_filtro("diagnostico"), description="Diagnosis:", layout=widgets.Layout(width="280px"))
m1_sexo = widgets.Dropdown(options=opciones_filtro("sexo"), description="Gender:", layout=widgets.Layout(width="200px"))
m1_estado_inv = widgets.Dropdown(options=opciones_filtro("estado_investigacion"), description="Research Status:", layout=widgets.Layout(width="260px"))
m1_con_imagenes = widgets.Checkbox(value=False, description="Only with images")
m1_con_exoma = widgets.Checkbox(value=False, description="Only with exome")
m1_boton_buscar = widgets.Button(description="Search", button_style="primary", icon="search")
m1_boton_limpiar = widgets.Button(description="Clear filters", icon="eraser")

m1_selector = widgets.Select(options=[], description="Select:", layout=widgets.Layout(width="500px"), rows=8)
m1_tabla_out = widgets.Output()
m1_ficha_out = widgets.Output()
m1_contador = widgets.HTML(value="<i>No search performed yet.</i>")

m1_resultados_actuales = {"df": df_master.iloc[0:0]}

def m1_ejecutar_busqueda(_=None):
    resultado = buscar_pacientes(
        df_master,
        patno_texto=m1_patno.value,
        alias_texto=m1_alias.value,
        estado=m1_estado.value,
        diagnostico=m1_diagnostico.value,
        sexo=m1_sexo.value,
        estado_investigacion=m1_estado_inv.value,
        solo_con_imagenes=m1_con_imagenes.value,
        solo_con_exoma=m1_con_exoma.value,
    )
    m1_resultados_actuales["df"] = resultado
    m1_contador.value = f"<b>{len(resultado)}</b> patients found."

    with m1_tabla_out:
        clear_output()
        cols = columnas_tabla_resumen()
        display(resultado[cols].reset_index(drop=True))

    etiquetas = []
    for _, fila in resultado.iterrows():
        patno_v = valor_o_na(fila, "patno")
        alias_v = valor_o_na(fila, "alias")
        etiqueta = f"{patno_v} - {alias_v}" if alias_v != "Not available" else f"{patno_v}"
        etiquetas.append((etiqueta, fila[COL["patno"]]))
    m1_selector.options = etiquetas

    with m1_ficha_out:
        clear_output()


def m1_limpiar(_=None):
    m1_patno.value = ""
    m1_alias.value = ""
    m1_estado.value = "Todos"
    m1_diagnostico.value = "Todos"
    m1_sexo.value = "Todos"
    m1_estado_inv.value = "Todos"
    m1_con_imagenes.value = False
    m1_con_exoma.value = False
    m1_ejecutar_busqueda()


def m1_mostrar_ficha(change):
    if change["new"] is None:
        return
    patno_sel = change["new"]
    fila = df_master[df_master[COL["patno"]] == patno_sel].iloc[0]
    with m1_ficha_out:
        clear_output()
        display(HTML(generar_ficha_html(fila, modo_completo=False)))


m1_boton_buscar.on_click(m1_ejecutar_busqueda)
m1_boton_limpiar.on_click(m1_limpiar)
m1_selector.observe(m1_mostrar_ficha, names="value")
```

```python
fila_filtros_1 = widgets.HBox([m1_patno, m1_alias, m1_estado])
fila_filtros_2 = widgets.HBox([m1_diagnostico, m1_sexo, m1_estado_inv])
fila_filtros_3 = widgets.HBox([m1_con_imagenes, m1_con_exoma])
fila_botones = widgets.HBox([m1_boton_buscar, m1_boton_limpiar])

modo1 = widgets.VBox([
    widgets.HTML("<h3>Explore all patients</h3>"),
    fila_filtros_1,
    fila_filtros_2,
    fila_filtros_3,
    fila_botones,
    m1_contador,
    m1_tabla_out,
    widgets.HTML("<hr><b>Select a patient to view their card:</b>"),
    m1_selector,
    m1_ficha_out,
])

# Initial search showing all patients
m1_ejecutar_busqueda()
```

## 5. Mode 2 — Complete Patients (images + exome)

```python
df_completos = df_master[df_master["_paciente_completo_bool"]].copy()
print("Pacientes completos (imágenes + exoma):", len(df_completos))
```

```python
m2_patno = widgets.Text(placeholder="Search by PATNO", description="PATNO:", layout=widgets.Layout(width="250px"))
m2_alias = widgets.Text(placeholder="Search by Alias (SI)", description="Alias:", layout=widgets.Layout(width="250px"))
m2_orden = widgets.Dropdown(
    options=[c for c in ["patno", "edad", "alias"] if COL.get(c)],
    description="Sort by:",
    layout=widgets.Layout(width="250px"),
)
m2_boton_buscar = widgets.Button(description="Search", button_style="primary", icon="search")
m2_boton_listar = widgets.Button(description="List all", icon="list")

m2_selector = widgets.Select(options=[], description="Select:", layout=widgets.Layout(width="500px"), rows=8)
m2_tabla_out = widgets.Output()
m2_ficha_out = widgets.Output()
m2_contador = widgets.HTML(value="<i>No search performed yet.</i>")


def m2_ejecutar_busqueda(_=None):
    resultado = df_completos.copy()

    if m2_patno.value.strip():
        resultado = resultado[resultado[COL["patno"]].astype(str).str.contains(m2_patno.value.strip(), na=False)]

    if m2_alias.value.strip() and COL.get("alias"):
        resultado = resultado[resultado[COL["alias"]].astype(str).str.contains(m2_alias.value.strip(), case=False, na=False)]

    columna_orden = COL.get(m2_orden.value)
    if columna_orden:
        resultado = resultado.sort_values(columna_orden)

    m2_contador.value = f"<b>{len(resultado)}</b> complete patients found."

    with m2_tabla_out:
        clear_output()
        cols = columnas_tabla_resumen()
        display(resultado[cols].reset_index(drop=True))

    etiquetas = []
    for _, fila in resultado.iterrows():
        patno_v = valor_o_na(fila, "patno")
        alias_v = valor_o_na(fila, "alias")
        etiqueta = f"{patno_v} - {alias_v}" if alias_v != "Not available" else f"{patno_v}"
        etiquetas.append((etiqueta, fila[COL["patno"]]))
    m2_selector.options = etiquetas

    with m2_ficha_out:
        clear_output()


def m2_listar_todos(_=None):
    m2_patno.value = ""
    m2_alias.value = ""
    m2_ejecutar_busqueda()


def m2_mostrar_ficha(change):
    if change["new"] is None:
        return
    patno_sel = change["new"]
    fila = df_completos[df_completos[COL["patno"]] == patno_sel].iloc[0]
    with m2_ficha_out:
        clear_output()
        display(HTML(generar_ficha_html(fila, modo_completo=True)))


m2_boton_buscar.on_click(m2_ejecutar_busqueda)
m2_boton_listar.on_click(m2_listar_todos)
m2_selector.observe(m2_mostrar_ficha, names="value")
```

```python
modo2 = widgets.VBox([
    widgets.HTML("<h3>Patients ready for research (images + exome)</h3>"),
    widgets.HBox([m2_patno, m2_alias, m2_orden]),
    widgets.HBox([m2_boton_buscar, m2_boton_listar]),
    m2_contador,
    m2_tabla_out,
    widgets.HTML("<hr><b>Select a patient to view their card:</b>"),
    m2_selector,
    m2_ficha_out,
])

# Initial listing of all complete patients
m2_ejecutar_busqueda()
```

## 6. Final interface assembly (tabs)

```python
tabs = widgets.Tab(children=[modo1, modo2])
tabs.set_title(0, "All patients")
tabs.set_title(1, "Patients with images + exome")

display(widgets.HTML("<h2>PPMI Explorer — Master Patient Index</h2>"))
display(tabs)
```
